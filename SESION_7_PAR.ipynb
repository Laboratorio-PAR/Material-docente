### 1. Programación funcional.

La **programación funcional** se basa principalmente en el uso de funciones, siendo estas (casi) la única herramienta que el lenguaje ofrece. Por ello, en lenguajes puramente funcionales como *Haskell* no existen bucles `for` o `while`.

A pesar de que Python no es un lenguaje puramente funcional, dispone de algunas primitivas propias de lenguajes funcionales, como `map`, `filter` y `reduce`. Todas ellas ofrecen una alternativa al uso de bucles para resolver ciertos problemas.

#### 1.1 Función `map`.

La función ` map() ` es una utilidad integrada en Python que aplica una función especificada a cada elemento de un iterable (como una lista, una tupla o una cadena) y devuelve un iterador con los resultados transformados.
Como resultado, promueve la inmutabilidad y la reutilización del código.

<img src="./images/map.jpg" width="500">

````python
# Ejemplo de uso de la función map()
lista = [1, 2, 3]
lista_doble = map(lambda x: x * 2, lista)
print(list(lista_doble))
````

#### 1.2 Función `filter`.

La función `filter()` permite la extracción de elementos de una lista, tupla o cualquier otro iterable, teniendo en cuenta que cumplan una determinada condición de **filtrado**.

Su sintaxis es la siguiente:
``` python
filter(function, iterable)
```
dónde:
- *function* es una prueba/condición que se aplica a cada elemento del iterable y que será `True`o `False`.
- *iterable* es una lista, tupla, cadena o cualquier tipo de objeto sobre el que se pueda iterar.

La función `filter()`retorna un objeto de tipo *filter*, por lo que es necesario 'envolver' los resultados obtenidos en una tupla o una lista.

````python
# Ejemplo de la función filter()
lista = [7, 6, 10, 13, 1, 24]
pares = filter(lambda x: x % 2 == 0, lista)
print(list(pares))
print(type(pares)) # tipo de la variable pares
````

#### 1.3 Función `reduce`.

La función `reduce()` permite la reducción de todos los elementos de una lista, tupla o cualquier otro iterable que se tenga como entrada de la función a un único valor después de haber aplicado un criterio concreto.

Para usar la función `reduce()` es necesario importarla desde la biblioteca `functools`, de la siguiente forma
``` python
from functools import reduce
```

````python
# Ejemplo de uso de la función reduce()
from functools import reduce
lista = [1, 2, 3, 4, 5]
suma = reduce(lambda x, y: x + y, lista)
print(suma)
````

````python
# Otra posible solución usando add
from functools import reduce
from operator import add
lista = [1, 2, 3, 4, 5]
suma = reduce(add, lista)
print(suma)
````

````python
# Si aplicamos la multiplicación
from functools import reduce
from operator import mul
lista = [1, 2, 3, 4, 5]
producto = reduce(mul, lista)
print(producto)
````

````python
# Ahora la multiplicación con una función lambda
from functools import reduce
lista = [1, 2, 3, 4, 5]
producto = reduce(lambda x, y: x * y, lista)
print(producto)
````

### 2. Profundizando en la programación funcional.

Después de revisar el comportamiento de `map`, `filter` y `reduce`, se puede intuir cuales son los principios básicos de la programación funcional, de los que cabe destacar los siguientes:

- Los lenguajes puramente funcionales no usan bucles `for` y `while` (Python no es puramente funcional &#9940;)
- Las funciones son ciudadanos de **primera clase**, lo que implica que la mayor parte del código se basa en funciones.
- En la programación funcional se da prioridad a las ***funciones puras***, que son aquellas que no tienen *efectos secundarios*. (Las funciones puras son aquellas que no dependen de variables externas/globales, por lo que para las mismas entradas se generan siempre las mismas salidas)
- En la programación funcional se suelen usar normalmente **variables inmutables** (no se pueden modificar después de haberse creado)

Destacar que, a pesar de que Python no es un lenguaje puramente funcional, dispone de las anteriores funciones que son propias de ese tipo de lenguajes.

A continuación se muestra un ejemplo de aplicación de programación funcional, tomando los datos de varias personas almacenados en un diccionario y realizando aplicando funciones hasta llegar al cálculo de la media de edad de los varones:

````python
from functools import reduce
personas = [
    {'Nombre': 'Juan', 'Edad': 25, 'Sexo': 'M'},
    {'Nombre': 'Ana', 'Edad': 22, 'Sexo': 'F'},
    {'Nombre': 'Carlos', 'Edad': 28, 'Sexo': 'M'},
    {'Nombre':'Patricia', 'Edad': 23, 'Sexo': 'F'},
    {'Nombre': 'Maria', 'Edad': 21, 'Sexo': 'F'},
    {'Nombre': 'Jose', 'Edad': 24, 'Sexo': 'M'},
]
varones = list(filter(lambda x: x['Sexo'] == 'M', personas))
suma_edades = reduce(lambda x, y: x + y['Edad'], varones, 0)
media_edad = suma_edades / len(varones)
print(media_edad)
````

````python
media_edad = reduce(lambda suma, p: suma + p['Edad'], filter(lambda x: x['Sexo']=='M',personas), 0) / len(list(filter(lambda x: x['Sexo']=='M',personas)))
print(media_edad)
````

### 3. Iterar usando la función `zip`.

La función `zip()` toma cualquier número de objetos iterables y devuelve un objeto iterador de tuplas. Es decir, si se le pasa dos listas como entrada, lo que se obtiene como resultado es una tupla con pares de elementos, donde cada par estará formado por un elemento de cada lista de entrada.

````python
# Ejemplo de la función zip()
lista1 = [23, 24, 28]
lista2 =['Juan', 'Ana', 'Sonia']
lista_personas = zip(lista1, lista2)
print(list(lista_personas))
````

````python
# Ejemplo en el que se muestra solo contenido del zip
series = ['Smallville', 'WestWorld', 'Game of Thrones']
directores = ['James Marshall', 'Jonathan Nolan', 'Alan Taylor']
anio = ['2001', '2016', '2011']

for a, b, c in  zip(series, directores, anio):
    print(f' Serie: {a} \n dirigida por: {b} \n Estreno: {c}')
````

Existe la posibilida de 'separar' una lista usando el `*` de la siguiente forma:

````python
lista_zipeada = [(1,4),(2,5),(3,6)]
lista_u, lista_v = zip(*lista_zipeada)
print("Lista u: ", lista_u)
print("Lista v: ", lista_v)
````

# Función anónima.

Una función anónima (o lambda) es una función sin nombre, definida en una sola línea con la palabra clave lambda.
La sintaxis es

``` python
lambda argumentos: expresión
```
Es equivalente a escribir una función normal, pero de forma compacta:

````python
# Función normal
def doble(x):
    return x * 2

# Equivalente como lambda
doble = lambda x: x * 2
````

Aunque se pueden asignar a una variable (como en el ejemplo anterior), su uso más natural es pasarlas directamente como argumento a otra función, sin necesidad de darles nombre:

````python
numeros = [3, 1, 4, 1, 5, 9, 2]

# Ordenar una lista de tuplas por el segundo elemento
pares = [(1, 'b'), (3, 'a'), (2, 'c')]
print(list(sorted(pares, key=lambda x: x[1])))   # → [(3,'a'), (1,'b'), (2,'c')]

# Filtrar solo los pares
print(list(filter(lambda x: x % 2 == 0, numeros))) # → [4, 2]

# Multiplicar cada elemento por 3
print(list(map(lambda x: x * 3, numeros)))  # → [9, 3, 12, 3, 15, 27, 6]
````

Limitaciones importantes respecto a una función normal:

- Solo puede contener una expresión (no sentencias como if, for, return…)
- No admite anotaciones de tipo
- Dificulta la legibilidad si se abusa de ellas

La guía de estilo de Python (**PEP 8**) recomienda no asignarlas a variables — si se necesita reutilizarla, mejor definir una función normal con `def`.

***Las lambdas 'brillan' cuando la lógica es simple y de un solo uso.***

````python
f = lambda x : x % 2 == 0
print(f(5)) # False
print(f(4))
````

# Operador ternario.

Los operadores ternarios, también conocidos como expresiones condicionales, evalúan si una expresión es verdadera o no. Se añadieron a Python en la versión 2.4 y su sintaxis es la siguiente:
``` python
condicion_si_es_verdadero if condicion else condicion_si_es_falso
```
Ejemplo:

````python
es_correcto = True

situacion = "Correcto" if es_correcto else "Incorrecto"
print(situacion)
````

Otra forma de uso menos habitual es la siguiente:

```python
(valor_falso_test, valor_verdadero_test)[test]
```
Ejemplo:

````python
es_correcto = True
situacion =("Incorrecto", "Correcto")[es_correcto]
print(situacion)
````

# List comprehension
- Una forma sucinta de generar listas o colecciones
- Es **"Pythonic"**
- https://python-course.eu/python3_list_comprehension.php

````python
# crear una lista transformando valores de otra con for loops
final_values = []

for x in range(-4,5):
    final_values.append(x**2)

print(final_values)
````

````python
# con list comprehension
final_values = [x**2 for x in range(-4,5)]
print(final_values)
````

### Forma general
[<expresión1> if <condición\> else <expresión2> for value in <colección\>]
- __Expresión1 / expresión2__: cada uno de los elementos que conformará la lista
- __Condición__: escoge entre expresión1 y expresión2
- __for__ loop: generador de la secuencia

````python
print([x**2 for x in range(10)])
````

````python
print([s.upper() for s in ['copos', 'avena', 'leche']])
````

````python
print([2 * i for i in range(1, 6) if i % 2 == 0])
````

````python
notas = {'Rosa':10, 'Paco':7.5, 'Pedro':2, 'Natalia':5, 'Silvia':4, 'Juan':8}
print("Aprobados: ", [nombre for nombre, nota in notas.items() if nota >= 5])
````

````python
# Eleva al cuadrado los elementos positivos y divide por 2 los negativos,
# generando una nueva lista
list1 = [x**2 if x > 0 else x/2 for x in range(-4,5)]
print(list1)
````

````python
# Si no hay ningún resto = 0 entre sus divisores, el número es primo!
print([8 %d for d in range(2, 8)]) # Devuelve los restos de dividir 8 entre 2, 3, 4, 5, 6 y 7 --> no es primo
print([17 %d for d in range(2, 17)]) # Si es primo.
````

````python
[i for i in range(100000) if i == 1000][0]
````

````python
next(i for i in range(100000) if i == 1000)
````

````python
# Filtrar elementos con una condición
pares = [x for x in range(10) if x % 2 == 0]
print("Números pares:", pares)  # [0, 2, 4, 6, 8]
````

````python
# similar para diccionarios y sets, pero con {}
items = ['Banana','Pear','Olives']
price = [1.1,1.4,2.4]
shopping = {k:v for k,v in zip(items,price)}
print(shopping)
````

````python
items = ['Banana','Pear','Olives']
price = [1.1,1.4,2.4]
for k in zip(items, price):
    print(type(k))
````

````python
# Anidamiento de List Comprehensions (Generar una tabla de multiplicar)
# Se generan tres numeros para cada variable x e y (range(1,4) genera [1,2,3])

tabla_multiplicar = [x * y for x in range(1, 4) for y in range(1, 4)]
print("Tabla de multiplicar:", tabla_multiplicar)  # [1, 2, 3, 2, 4, 6, 3, 6, 9]
````

### Usando diccionarios con comprehension.

````python
# Ejemplo 1: doblar cada valor del diccionario
dict1 = {'a': 1, 'b': 2, 'c': 3, 'd': 4, 'e': 5}
double_dict1 = {key:value * 2 for (key, value) in dict1.items()}
print(double_dict1) # {'a': 2, 'c': 6, 'b': 4, 'e': 10, 'd': 8}
````

````python
# Ejemplo 2: transformar de grados Fahrenheit a Celsius
# Inicializamos el diccionario en grados Fahrenheit
fahrenheit = {'t1':-30, 't2':-20, 't3':-10, 't4': 0}
# Obtenemos los valores correspondientes en grados celsius y creamos el nuevo diccionario
celsius = {k:(float(5) / 9) * (v - 32) for (k, v) in fahrenheit.items()}
print(celsius)
````

````python
# Ejemplo 3: elevar a la potencia 2 cada valor del diccionario: Obtener diccionario de cuadrados.
numeros_cuadrados = {x: x**2 for x in range(5)} # el range(5) genera [0,1,2,3,4]
print("Diccionario de cuadrados:", numeros_cuadrados)  # {0: 0, 1: 1, 2: 4, 3: 9, 4: 16}
````
