## 2.- Orientación a eventos
    • Permite crear aplicaciones gráficas de escritorio o móvil.
    • Es lo contrario a programación secuencial, donde:
        - Existe un único flujo de ejecución y estado.
    • No existe un único flujo de ejecución e interacción.
    • Atiende a sucesos/eventos (señales) que ocurren y desencadenan métodos de nuestro código.
    • Funcionamiento asíncrono (concurrencia, hilos…).
    • ¿Cómo se programa? Callbacks + orientación a objetos.
        - Se especifican las señales a los que debe responder el programa.
        - Se especifican las acciones a realizar cuando ocurren las señales.

## 6.- Python Avanzado.

• `app.MainLoop()` está en un bucle que solo reacciona cuando hay eventos por procesar.

• Cuando ocurre una evento en la GUI, se emite una señal desde el elemento (widget) donde se ha producido. Entonces se procesa, o se pone a la cola de eventos por procesar --> `callback`

• Mientras haya eventos/señales en la cola, se procesan.

• Cuando se procesa una señal, el control pasa al método asociado de `callback` (programación funcional).

• Cada vez que creamos un widget, debemos especificar al menos el primer parámetro, que normalmente es el widget que lo contiene (widget padre, `parent`): `None` si no tiene `parent`.

• Existen señales que los widgets heredan `(EVT_CLOSE)`.

• Existen señales específicas de los widgets [wx.EVT_BUTTON, wx.EVT_MENU…].

• Se asocian (registran) señales a métodos de `callback` mediante el método Bind, uniendo el widget en el que se produce el evento de la GUI con su método de `callback`.
``` python
    miContenedor.Bind(señal, metodoCallback, widgetSource)
    def metodoCallback(evento):
```

• Siempre deberemos poner el argumento evento en nuestros métodos callback.
``` python
print(event.getID()) # Para identificar el id de miContenedor
```
• Deberemos indicar el `widgetSource` solo cuando tengamos varios elementos gráficos del **mismo tipo** dentro de un mismo widget contenedor, y queramos diferenciarlos en un *mismo* metodoCallback. En caso de solo tener uno, podemos tanto ponerlo como no. Dentro de nuestro método callback podemos hacer:
